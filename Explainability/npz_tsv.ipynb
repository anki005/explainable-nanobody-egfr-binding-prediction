```python
import pickle
import numpy as np
import pandas as pd

with open(r"C:\Users\ankit\Downloads\good_positive_explanations_90_test.pkl", "rb") as f:
    positive_explanations = pickle.load(f)
all_explanations = positive_explanations  
npz_dict = {}
metadata_rows = []

for i, rec in enumerate(all_explanations):
    key = f"exp_{i}"
    exp = np.array(rec["explanation_matrix"])
    npz_dict[key] = exp

    metadata_rows.append({
        "index": i,
        "npz_key": key,
        "Nanobody_id": rec["Nanobody_id"],
        "True_label": rec["True_label"],
        "Pred_label": rec["Pred_label"],
        "Pred_prob_1": rec["Pred_prob_1"],
        "Sequence": rec["Sequence"],
        "Round": rec["Round"],
        "Aligned_sequence": rec["Aligned_sequence"],
        "CDRs": rec["CDRs"],
    })
np.savez_compressed(r"C:\Users\ankit\Downloads\positive_explanations_90_test.npz", **npz_dict)  
metadata_df = pd.DataFrame(metadata_rows)
metadata_df.to_csv(r"C:\Users\ankit\Downloads\positive_explanations_90_test.tsv", sep="\t", index=False)
```
