```python
import pandas as pd
from sklearn.model_selection import StratifiedGroupKFold
```

```python
val_df = pd.read_csv(r"D:\Thesis EGFR Round 3\Extra\changed_final_val_seqs_new.csv")
```

```python
x_val = val_df.drop(columns=['Label'])
y_val = val_df['Label']
group = val_df['Cluster_number']

sgkf = StratifiedGroupKFold(n_splits=2, shuffle=True, random_state=42)
for va1_idx, va2_idx in sgkf.split(x_val, y_val, group):
    val1_df = val_df.iloc[va1_idx].copy()
    val2_df = val_df.iloc[va2_idx].copy()
    break
```

```python
print("Original:")
print(val_df["Label"].value_counts(normalize=True))

print("\nPart 1:")
print(val1_df["Label"].value_counts(normalize=True))
print(val1_df["Cluster_number"].nunique())

print("\nPart 2:")
print(val2_df["Label"].value_counts(normalize=True))
print(val2_df["Cluster_number"].nunique())

print("\nCluster overlap:",
      set(val1_df["Cluster_number"]).intersection(set(val2_df["Cluster_number"])))
```

```python
val1_df.to_csv(r"D:\Thesis EGFR Round 3\Extra\changed_final_val_seqs_new_validation.csv", index=False)
val2_df.to_csv(r"D:\Thesis EGFR Round 3\Extra\changed_final_val_seqs_new_test.csv", index=False)
```
