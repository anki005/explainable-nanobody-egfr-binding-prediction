New method for check and removal for duplicates from rounds 

```python
r1 = set(final_df_unique['igghAaSequence'])
r2 = set(final_df2['igghAaSequence'])
r3 = set(final_df3_unique['igghAaSequence'])
```

```python
def count_patterns(r1, r2, r3):
    return {
        "only_round1": len(r1 - r2 - r3),
        "only_round2": len(r2 - r1 - r3),
        "only_round3": len(r3 - r1 - r2),
        "round1_2": len((r1 & r2) - r3),
        "round1_3": len((r1 & r3) - r2),
        "round2_3": len((r2 & r3) - r1),
        "round1_2_3": len(r1 & r2 & r3),
        "total_unique": len(r1 | r2 | r3),
    }
```

```python
def plot_venn(r1, r2, r3, title="Sequence overlap"):
    plt.figure(figsize=(6, 6))
    venn3([r1, r2, r3], set_labels=("Round 1", "Round 2", "Round 3"))
    plt.title(title)
    plt.show()
```

```python
def remove_and_assign(r1, r2, r3, seq_col="igghAaSequence"):
    df = pd.concat([
        r1.assign(_round=1),
        r2.assign(_round=2),
        r3.assign(_round=3),
    ], ignore_index=True)

    target = df.groupby(seq_col)["_round"].max()
    df["_target"] = df[seq_col].map(target)
    kept = df[df["_round"] == df["_target"]].copy()
    new_r1 = kept[kept["_round"] == 1].drop(columns=["_round", "_target"])
    new_r2 = kept[kept["_round"] == 2].drop(columns=["_round", "_target"])
    new_r3 = kept[kept["_round"] == 3].drop(columns=["_round", "_target"])

    return new_r1, new_r2, new_r3
```

```python
counts = count_patterns(r1, r2, r3)
print(counts)
plot_venn(r1, r2, r3, title="Nanobody EGFR Sequence Overlap")
new_r1, new_r2, new_r3 = remove_and_assign(final_df_unique, final_df2, final_df3_unique, seq_col="igghAaSequence")
```

Add Unique ID to the dataset  

```python

def nanobody_id(sequence):
    seq = str(sequence).strip().upper().replace(" ", "")
    hash_hex = hashlib.sha256(seq.encode()).hexdigest()[:12]
    return f"Nb_{hash_hex}"

combined_df = pd.read_csv(r"D:\Thesis EGFR Round 3\Dataset\EG-comnined_nonredundant_unique.csv")

print(combined_df.columns)
combined_df["Nanobody_id"] = combined_df["igghAaSequence"].apply(nanobody_id)
combined_df.to_csv(r"D:\Thesis EGFR Round 3\Dataset\EG-combined dataset with ID.csv", index=False)
```
