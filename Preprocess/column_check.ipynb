Check the sequences with insertions

```python
check_df = pd.read_csv(r"D:\Thesis EGFR Round 3\Dataset\ANARCII\EG-combined ANARCII_imgt_wide_fixed.csv")
positions = ["6B", "6C", "6D", "7B", "7C", "9B", "9C","32C", "32D", "33C", "33D", "42B", "42C", "42D", "79B", "79C", "119B"]    

output_rows = []
for _, row in check_df.iterrows():
    nb_id = row["Name"]
    inserts = []
    for p in positions:
        col = str(p)      
        if col in check_df.columns:
            aa = row[col]
            if aa not in ["", "-", None]:   
                inserts.append(col)

    if inserts:
        output_rows.append({"Name": nb_id, "Insertions": ", ".join(inserts)})

output_df = pd.DataFrame(output_rows)
output_df.to_csv(r"D:\Thesis EGFR Round 3\Dataset\ANARCII\insertions.csv", index=False)
```

Remove the sequences with insertions and delete respective column

```python
insertions = pd.read_csv(r"D:\Thesis EGFR Round 3\Dataset\ANARCII\insertions.csv")
check_df = pd.read_csv(r"D:\Thesis EGFR Round 3\Dataset\ANARCII\EG-combined ANARCII_imgt_wide_fixed.csv")
df_clean = check_df[~check_df["Name"].isin(insertions["Name"])]
print(len(df_clean))

df_clean = df_clean.drop(columns=["6B", "6C", "6D", "7B", "7C", "9B", "9C","32C", "32D", "33C", "33D", "42B", "42C", "42D", "79B", "79C", "119B"])
df_clean.to_csv(r"D:\Thesis EGFR Round 3\Dataset\ANARCII\EG-ANARCII_imgt_wide_fixed_filtered.csv", index=False)
```
