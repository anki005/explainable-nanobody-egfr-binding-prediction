CDRs extraction for clustering

```python
cdrs_df = pd.read_csv(r"D:\Thesis EGFR Round 3\Dataset\final_filtered_seqs_merged.csv")
cdrs_df = cdrs_df[['Nanobody_id','CDR1','CDR2','CDR3', 'Label']]
cdrs_df["Concatenated_CDRs"] = cdrs_df["CDR1"] + cdrs_df["CDR2"] + cdrs_df["CDR3"]
cdrs_df = cdrs_df[["Nanobody_id", "Concatenated_CDRs", "Label"]]
print(len(cdrs_df))
cdrs_df.to_csv(r"D:\Thesis EGFR Round 3\Dataset\concatenated_cdrs.csv", index=False)
```
